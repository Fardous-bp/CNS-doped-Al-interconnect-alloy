```python
from ase import units
from ase.md.langevin import Langevin
from ase.md.velocitydistribution import MaxwellBoltzmannDistribution, Stationary, ZeroRotation
from ase.md import MDLogger
from ase.io import Trajectory
import time

# Copy the relaxed 6.25% Al supercell
md_atoms_6_25 = atoms_super_6_25.copy()
md_atoms_6_25.calc = calculator_pbe

# Initialize velocities at 473 K
MaxwellBoltzmannDistribution(md_atoms_6_25, temperature_K=473)
Stationary(md_atoms_6_25)
ZeroRotation(md_atoms_6_25)

# MD parameters
temperature_K = 473
timestep = 1.0 * units.fs
friction = 0.01 / units.fs

dyn = Langevin(
    md_atoms_6_25,
    timestep,
    temperature_K=temperature_K,
    friction=friction,
    fixcm=False
)

# Output files
traj_file = "CuNi2Sn_6.25Al_2x2x2_MD_2ps.traj"
log_file = "CuNi2Sn_6.25Al_2x2x2_MD_2ps.log"

traj = Trajectory(traj_file, "w", md_atoms_6_25)
dyn.attach(traj.write, interval=10)

logger = MDLogger(
    dyn,
    md_atoms_6_25,
    log_file,
    header=True,
    stress=False,
    peratom=False,
    mode="w"
)
dyn.attach(logger, interval=10)

# Run 2 ps
start_time = time.time()

dyn.run(2000)

elapsed = time.time() - start_time

traj.close()

print("\n=== MD COMPLETED ===")
print("Total steps:", 2000)
print("Simulation time: 2 ps")
print("Number of atoms:", len(md_atoms_6_25))
print(f"Runtime: {elapsed/60:.2f} minutes")
print(f"Time per step: {elapsed/2000:.4f} s")
print("Trajectory:", traj_file)
print("Log:", log_file)
```

```python
from ase.io import read
import numpy as np

# Load trajectory
traj_6_25 = read(
    "CuNi2Sn_6.25Al_2x2x2_MD_2ps.traj",
    index=":"
)

dt_ps = 0.01  # trajectory saved every 10 fs = 0.01 ps

# ---------------------------------------------------------
# Build unwrapped coordinates using ASE's minimum-image
# displacement between consecutive frames
# ---------------------------------------------------------
wrapped = np.array([atoms.get_positions() for atoms in traj_6_25])
unwrapped = np.zeros_like(wrapped)
unwrapped[0] = wrapped[0]

for i in range(1, len(traj_6_25)):
    dr = wrapped[i] - wrapped[i-1]

    # Convert Cartesian displacement to fractional coordinates
    frac_dr = np.linalg.solve(
        traj_6_25[i-1].cell.array.T,
        dr.T
    ).T

    # Minimum-image displacement
    frac_dr -= np.round(frac_dr)

    # Convert back to Cartesian
    cart_dr = frac_dr @ traj_6_25[i-1].cell.array

    unwrapped[i] = unwrapped[i-1] + cart_dr


# ---------------------------------------------------------
# Time-origin averaged MSD
# ---------------------------------------------------------
max_lag = len(traj_6_25) - 1

msd = []
times = []

for lag in range(1, max_lag + 1):

    displacements = (
        unwrapped[lag:] - unwrapped[:-lag]
    )

    squared_displacements = np.sum(
        displacements**2,
        axis=2
    )

    msd_value = np.mean(squared_displacements)

    msd.append(msd_value)
    times.append(lag * dt_ps)

msd = np.array(msd)
times = np.array(times)


# ---------------------------------------------------------
# Print selected lag times
# ---------------------------------------------------------
print("\n=== 6.25% Al PBC-CORRECTED MSD ===")

for t_target in [0.25, 0.50, 0.75, 1.00, 1.25, 1.50, 1.75, 2.00]:

    idx = np.argmin(np.abs(times - t_target))

    print(
        f"{times[idx]:5.2f} ps : "
        f"{msd[idx]:.6f} Å²"
    )
```

```python
from ase.io import read
import numpy as np

traj = read(
    "CuNi2Sn_6.25Al_2x2x2_MD_2ps.traj",
    index=":"
)

dt_ps = 0.01

# ---------------------------------------------------------
# Unwrap trajectories using minimum-image displacements
# ---------------------------------------------------------
wrapped = np.array([atoms.get_positions() for atoms in traj])
unwrapped = np.zeros_like(wrapped)
unwrapped[0] = wrapped[0]

for i in range(1, len(traj)):
    dr = wrapped[i] - wrapped[i-1]

    frac_dr = np.linalg.solve(
        traj[i-1].cell.array.T,
        dr.T
    ).T

    frac_dr -= np.round(frac_dr)

    cart_dr = frac_dr @ traj[i-1].cell.array

    unwrapped[i] = unwrapped[i-1] + cart_dr


# ---------------------------------------------------------
# Species masks
# ---------------------------------------------------------
symbols = np.array(traj[0].get_chemical_symbols())

species = ["Al", "Cu", "Ni", "Sn"]

masks = {
    sp: symbols == sp
    for sp in species
}


# ---------------------------------------------------------
# Species-resolved time-origin averaged MSD
# ---------------------------------------------------------
target_times = [0.25, 0.50, 0.75, 1.00,
                1.25, 1.50, 1.75, 2.00]

print("\n=== 6.25% Al SPECIES-RESOLVED PBC-CORRECTED MSD ===")
print("MSD units: Å²\n")

header = "Time (ps)"
for sp in species:
    header += f"{sp:>12}"
print(header)

print("-" * len(header))

for t_target in target_times:

    lag = int(round(t_target / dt_ps))

    if lag >= len(traj):
        continue

    values = []

    for sp in species:

        mask = masks[sp]

        displacements = (
            unwrapped[lag:, mask, :]
            - unwrapped[:-lag, mask, :]
        )

        squared_displacements = np.sum(
            displacements**2,
            axis=2
        )

        msd_value = np.mean(squared_displacements)

        values.append(msd_value)

    print(
        f"{t_target:8.2f}"
        + "".join(f"{v:12.6f}" for v in values)
    )
```

```python
from ase.io import Trajectory
import numpy as np
import matplotlib.pyplot as plt

traj = Trajectory("CuNi2Sn_6.25Al_2x2x2_MD_2ps.traj")

nframes = len(traj)
natoms = len(traj[0])

# ---------------------------------------------------------
# Build PBC-corrected unwrapped Cartesian positions
# ---------------------------------------------------------
positions = np.zeros((nframes, natoms, 3))

positions[0] = traj[0].get_positions()

for i in range(1, nframes):

    # Cartesian displacement between consecutive frames
    dr = traj[i].get_positions() - traj[i-1].get_positions()

    # Convert Cartesian displacement to fractional coordinates
    frac_dr = np.linalg.solve(
        traj[i-1].cell.array.T,
        dr.T
    ).T

    # Apply minimum-image convention
    frac_dr -= np.round(frac_dr)

    # Convert back to Cartesian coordinates
    cart_dr = frac_dr @ traj[i-1].cell.array

    # Accumulate unwrapped positions
    positions[i] = positions[i-1] + cart_dr


# ---------------------------------------------------------
# Time-origin averaged MSD
# ---------------------------------------------------------
msd = np.zeros(nframes)

for lag in range(1, nframes):

    displacements = positions[lag:] - positions[:-lag]

    squared = np.sum(displacements**2, axis=2)

    msd[lag] = np.mean(squared)


# ---------------------------------------------------------
# Time axis
# ---------------------------------------------------------
time_ps = np.arange(nframes) * 0.01


# ---------------------------------------------------------
# Print selected values
# ---------------------------------------------------------
print("6.25 at.% Al — PBC-corrected Time-origin averaged MSD")
print("-------------------------------------------------------")

for ps in [0.25, 0.50, 0.75, 1.00,
           1.25, 1.50, 1.75, 2.00]:

    idx = int(round(ps / 0.01))

    print(f"{ps:4.2f} ps : {msd[idx]:.6f} Å²")


# ---------------------------------------------------------
# Smooth continuous-looking MSD curve
# ---------------------------------------------------------
# ---------------------------------------------------------
# Plot
# ---------------------------------------------------------
plt.figure(figsize=(7, 5))

plt.plot(
    time_ps,
    msd,
    linewidth=2
)

plt.xlabel("Time (ps)")
plt.ylabel("MSD (Å²)")
plt.title(
    "PBC-Corrected MSD — 6.25 at.% Al, 473 K"
)

# Fixed scale for comparison across all compositions
plt.xlim(0, 2.0)
plt.ylim(0, 0.16)

plt.tight_layout()
plt.show()
```

```python
from ase.io import Trajectory
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# Load trajectory
# ---------------------------------------------------------
traj = Trajectory("CuNi2Sn_6.25Al_2x2x2_MD_2ps.traj")

nframes = len(traj)
natoms = len(traj[0])

symbols = np.array(traj[0].get_chemical_symbols())
elements = ["Al", "Cu", "Ni", "Sn"]

# ---------------------------------------------------------
# Build PBC-corrected unwrapped Cartesian positions
# ---------------------------------------------------------
positions = np.zeros((nframes, natoms, 3))

positions[0] = traj[0].get_positions()

for i in range(1, nframes):

    dr = (
        traj[i].get_positions()
        - traj[i - 1].get_positions()
    )

    # Cartesian displacement -> fractional displacement
    frac_dr = np.linalg.solve(
        traj[i - 1].cell.array.T,
        dr.T
    ).T

    # Minimum-image correction
    frac_dr -= np.round(frac_dr)

    # Fractional -> Cartesian
    cart_dr = frac_dr @ traj[i - 1].cell.array

    # Accumulate unwrapped positions
    positions[i] = positions[i - 1] + cart_dr

# ---------------------------------------------------------
# Species indices
# ---------------------------------------------------------
species_indices = {
    elem: np.where(symbols == elem)[0]
    for elem in elements
}

# ---------------------------------------------------------
# Species-resolved time-origin averaged MSD
# ---------------------------------------------------------
msd_species = {
    elem: np.zeros(nframes)
    for elem in elements
}

for lag in range(1, nframes):

    displacements = (
        positions[lag:]
        - positions[:-lag]
    )

    squared = np.sum(
        displacements**2,
        axis=2
    )

    for elem in elements:

        idx = species_indices[elem]

        if len(idx) > 0:

            msd_species[elem][lag] = np.mean(
                squared[:, idx]
            )

# ---------------------------------------------------------
# Time axis
# 10 fs = 0.01 ps per saved frame
# ---------------------------------------------------------
time_ps = np.arange(nframes) * 0.01

# ---------------------------------------------------------
# Print selected values
# ---------------------------------------------------------
print("=== 6.25 at.% Al SPECIES-RESOLVED PBC-CORRECTED MSD ===")
print("MSD units: Å²")
print()

print(
    f"{'Time (ps)':>10}"
    f"{'Al':>12}"
    f"{'Cu':>12}"
    f"{'Ni':>12}"
    f"{'Sn':>12}"
)

print("-" * 58)

for ps in [
    0.25, 0.50, 0.75, 1.00,
    1.25, 1.50, 1.75, 2.00
]:

    idx = int(round(ps / 0.01))

    print(
        f"{ps:10.2f}"
        f"{msd_species['Al'][idx]:12.6f}"
        f"{msd_species['Cu'][idx]:12.6f}"
        f"{msd_species['Ni'][idx]:12.6f}"
        f"{msd_species['Sn'][idx]:12.6f}"
    )

# ---------------------------------------------------------
# Plot continuous species-resolved curves
# ---------------------------------------------------------
plt.figure(figsize=(7, 5))

# IMPORTANT: actually plot each species
for elem in elements:

    plt.plot(
        time_ps,
        msd_species[elem],
        label=elem,
        linewidth=2
    )

# ---------------------------------------------------------
# Axis labels and title
# ---------------------------------------------------------
plt.xlabel("Time (ps)")
plt.ylabel("MSD (Å²)")

plt.title(
    "PBC-Corrected Species-Resolved MSD — "
    "6.25 at.% Al, 473 K"
)

# ---------------------------------------------------------
# Fixed scale for comparison across all compositions
# ---------------------------------------------------------
plt.xlim(0, 2.0)
plt.ylim(0, 0.20)

# ---------------------------------------------------------
# Legend and layout
# ---------------------------------------------------------
plt.legend()

plt.tight_layout()
plt.show()
```

```python
from ase.io import Trajectory
import numpy as np
import matplotlib.pyplot as plt

traj = Trajectory("CuNi2Sn_6.25Al_2x2x2_MD_2ps.traj")

nframes = len(traj)
natoms = len(traj[0])

# ---------------------------------------------------------
# Build PBC-corrected unwrapped Cartesian positions
# ---------------------------------------------------------
positions = np.zeros((nframes, natoms, 3))

positions[0] = traj[0].get_positions()

for i in range(1, nframes):

    # Cartesian displacement between consecutive frames
    dr = traj[i].get_positions() - traj[i-1].get_positions()

    # Convert Cartesian displacement to fractional coordinates
    frac_dr = np.linalg.solve(
        traj[i-1].cell.array.T,
        dr.T
    ).T

    # Apply minimum-image convention
    frac_dr -= np.round(frac_dr)

    # Convert back to Cartesian coordinates
    cart_dr = frac_dr @ traj[i-1].cell.array

    # Accumulate unwrapped positions
    positions[i] = positions[i-1] + cart_dr


# ---------------------------------------------------------
# Time-origin averaged MSD
# ---------------------------------------------------------
msd = np.zeros(nframes)

for lag in range(1, nframes):

    displacements = positions[lag:] - positions[:-lag]

    squared = np.sum(displacements**2, axis=2)

    msd[lag] = np.mean(squared)


# ---------------------------------------------------------
# Time axis
# ---------------------------------------------------------
time_ps = np.arange(nframes) * 0.01


# ---------------------------------------------------------
# Print selected values
# ---------------------------------------------------------
print("6.25 at.% Al — PBC-corrected Time-origin averaged MSD")
print("-------------------------------------------------------")

for ps in [0.25, 0.50, 0.75, 1.00,
           1.25, 1.50, 1.75, 2.00]:

    idx = int(round(ps / 0.01))

    print(f"{ps:4.2f} ps : {msd[idx]:.6f} Å²")


# ---------------------------------------------------------
# Smooth continuous-looking MSD curve
# ---------------------------------------------------------
plt.figure(figsize=(7,5))

plt.plot(
    time_ps,
    msd,
    linewidth=2
)

plt.xlabel("Time (ps)")
plt.ylabel("MSD (Å²)")
plt.title("Mean Squared Displacement — 6.25 at.% Al, 473 K")

plt.tight_layout()
plt.show()
```
